### MultiFC Stats

```python
import os
import pandas as pd

multifc_train_df = pd.read_json(os.path.join(os.getcwd(), "multifc", "train.json"), lines=True)
multifc_test_df = pd.read_json(os.path.join(os.getcwd(), "multifc", "test.json"), lines=True)
multifc_val_df = pd.read_json(os.path.join(os.getcwd(), "multifc", "validation.json"), lines=True)
```

```python
train_label_counts = multifc_train_df['label'].value_counts()
test_label_counts = multifc_test_df['label'].value_counts()
val_label_counts = multifc_val_df['label'].value_counts()

multifc_stats = {}
multifc_stats["train"] = train_label_counts
multifc_stats["test"] = test_label_counts
multifc_stats["validation"] = val_label_counts
```

### FEVER Stats

```python
# count "NOT ENOUGH INFO" entries with evidence
def count_nei_with_evidence(df, dataset_name):
    count = df[
        (df['label'] == 'NOT ENOUGH INFO') &
        (
            df['evidence_text'].astype(str).str.strip().ne('') |
            df['evidence_url'].notna() & df['evidence_url'].astype(str).str.strip().ne('')
        )
    ].shape[0]
    print(f"{dataset_name}: {count} 'NOT ENOUGH INFO' entries with some evidence provided.")
    return count

# Count for FEVER
nei_fever_train = count_nei_with_evidence(fever_train_df, "FEVER Train")
nei_fever_val = count_nei_with_evidence(fever_val_df, "FEVER Validation")
nei_fever_test = count_nei_with_evidence(fever_test_df, "FEVER Test")
```
